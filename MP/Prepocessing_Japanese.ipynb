```python
import pandas as pd
```

```python
df = pd.read_csv('Files/VA_Records.csv')
df = df.drop(columns=['Unnamed: 0'])
```

```python
df
```

```python
df_Filtering = df[['Personnel_ID', 'Personnel_Name']].drop_duplicates()
```

```python
def countVowels(s):
    vowels = "aeiouAEIOU"
    vowel_count = 0

    invalids = "lxvqLXVQáéíóúàèìòùäëïöüâêîôûãõæœAEIOUÁÉÍÓÚÀÈÌÒÙÄËÏÖÜÂÊÎÔÛÃÕÆŒñçšžłćđŋđ"
    
    for char in s:
        if char in vowels:
            vowel_count += 1
        if char in invalids:
            return -999
            
    return vowel_count
```

```python
def countLetters(string):
    return len(''.join(string.split()))
```

```python
def countConsecutiveConsonants(s):
    vowels = 'aeiouAEIOU'

    
    # Initialize variables
    count = 0
    max_count = 0
    in_consonant_sequence = False
    
    # Iterate through each character in the string
    for char in s:
        # Check if the character is a consonant
        if char.isalpha() and char not in vowels:
            count += 1
            in_consonant_sequence = True
        else:
            # If a sequence ends, update max_count
            if in_consonant_sequence:
                max_count = max(max_count, count)
            count = 0  # Reset count for next sequence
            in_consonant_sequence = False
    
    # Final check in case the string ends with a consonant sequence
    if in_consonant_sequence:
        max_count = max(max_count, count)
    
    return max_count
```

```python
df_Filtering['Vowel_Count'] = df_Filtering.Personnel_Name.apply(countVowels)
df_Filtering['Letter_Count'] = df_Filtering.Personnel_Name.apply(countLetters)
df_Filtering['Vowel_Percentage'] = df_Filtering.Vowel_Count / df_Filtering.Letter_Count
df_Filtering['Consecutive_Consonants'] = df_Filtering.Personnel_Name.apply(countConsecutiveConsonants)
```

```python
df_Filtering = df_Filtering[df_Filtering.Consecutive_Consonants <= 3].sort_values('Consecutive_Consonants')
df_Filtering
```

```python
df_Filtering = df_Filtering[df_Filtering.Vowel_Percentage > 0.42].sort_values('Vowel_Percentage')
df_Filtering
```

```python
df = df[df.Personnel_ID.apply(lambda x : x in df_Filtering.Personnel_ID.values)]
df.to_csv('Files/VA_Records_JP.csv')
df
```
