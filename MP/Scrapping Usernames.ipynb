```python
INCREMENT = 24
MAX_PAGES = 1200 / INCREMENT

output = set()

for USER_RANGE in np.arange(0, INCREMENT*MAX_PAGES, INCREMENT):
    try:
        URL = f'https://myanimelist.net/users.php?q&loc=Philippines&agelow=0&agehigh=0&g&show={USER_RANGE}'
        soup = Scraper().scrape(URL)
    
        a_hrefs = [td.find('a') for td in soup.find_all('td', attrs={'class':'borderClass'})]
        
        output = output | {a.get('href').replace('/profile/', '') for a in a_hrefs}
    except:
        print(f'ERROR:{USER_RANGE}')
```

```python
pd.Series(list(output)).to_csv('Usernames.csv')
```
