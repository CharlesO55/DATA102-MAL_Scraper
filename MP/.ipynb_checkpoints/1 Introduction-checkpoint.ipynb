# A Recommender System for Anime Voice Actor Casting

Reccomender systems are powerful tools in observing patterns and generating reccomendations based of those observed patterns.

There are numerous approaches this project will show which are

1. Association Rule Learning
2. Collaborative Filtering with Item/Item and User/User Matrices
3. Collaboritive Filtering with Model Based Single Value Decomposition

## Why?
### Building a Team
Voice acting is a career path some seek to professionaly engage in, yet building the right team may be difficult.


There are issues on Actor availability affected by their work load, personal health, schedule and strikes (e.g. SAG-AFTRA vs Genshin Impact)

A reccomender system could help address these issues by recommending substitute actors with similar performances or also recommend work to Actors based on their work history.


To dive into specifics...

For Producers, this would be a tool to recommend Casts that synergize well together and share history.

You will be provided similarity scores of cast members [Collabortive] and a frequent item sets [Association Rule Leaning] of most common team compositions.

For Voice Actors, this could recommend other Projects and Series they could join based once more on their synergy with other staff.

You will be provided similarity scores of similar anime projects.

## The Data

Data was scraped from MyAnimeList Season, Anime, User Watch History, and People pages.

These were all used in other notebooks you may explore.

However with regards to this submission, only Anime and People are the relevant data.

### Anime
Anime contains information on
1. Anime ID
2. Ratings/Members/Favorites
3. Genres/Themes

### People 
People pages consist of either Voice Actors, Staff, or Mangakas.
1. Anime ID
2. Character/Role/Position
3. Ratings/Members/Favorites

## Challenges
The biggest challenge is the lack of proper data regarding Voice Acting. There's no information regarding vocal features such as pitch, speed, etc.

The only published data on MyAnimeList.net is information on their Character's Rating and the Features of the Anime Project they took part in.

Hence, we will be working an unknown that it could be a Character's personality, design, or traits instead that provides the ratings instead of the Voice Acting.

## Evaluation

### Precision
The metric used is Precision. 

This is because we are examining the predictions made. What percentage of there were True vs False Positives?

If False Positives are too high, then these predictions are not noteworthy.
