```python
from selenium.webdriver.chrome.service import Service
from selenium import webdriver  
from selenium.webdriver.common.by import By

import numpy as np
import pandas as pd
#from bs4 import BeautifulSoup
import time
import re
```

```python
USERNAMES = pd.read_csv('Users/Usernames.csv', index_col=0).values.flatten()
len(USERNAMES)
```

```python
DRIVER_PATH = "driver/chromedriver.exe"
driver = webdriver.Chrome(service=Service(DRIVER_PATH))
```

```python
URL = f"https://myanimelist.net/animelist/{USERNAMES[0]}?status=7"
driver.get(URL)

# SCROLL DOWN
pause = 0.5
lastHeight = driver.execute_script("return document.body.scrollHeight")

while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(pause)
    newHeight = driver.execute_script("return document.body.scrollHeight")
    if newHeight == lastHeight:
        break
    lastHeight = newHeight
```

```python
results = driver.find_elements(By.XPATH, "//tr[@class='list-table-data']")
[r.text for r in results]
```

```python
results = np.array([])

for r in results:
    output = dict()

    try:
        link = r.find_element(By.XPATH, ".//a[@class='link sort']").get_attribute('href')
        output['MAL_id'] = int(re.findall(r'anime/(\d+)', link)[0])
        output['Title'] = re.findall(r'anime/\d+/(.*)', link)[0]
    except:
        print('TITLE ERROR: SKIPPED')
        continue
    
    try:
        score = int(r.find_element(By.XPATH, ".//span[contains(@class, 'score-label')]").text)
    except:
        score = np.nan
    finally:
        output['Score'] = score

    try:
        output['Type'] = r.find_element(By.XPATH, ".//td[@class='data type']").text
    except:
        pass

    #results.append(output)
    print(output)
```
